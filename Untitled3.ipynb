```python
# Python Programming – Week 1
## Getting Started with Python

**Domain:** Finance

**Problem Statement:**
Many students struggle to track their daily expenses and manage their money.
I would like to develop a simple expense tracking system that helps students
record expenses, categorize spending, and understand their financial habits.

**Objective:** Learn Python fundamentals and build a foundation for AI and Data Science.
```

```python
# Integer
age = 20

# Float
monthly_budget = 5000.50

# String
student_name = "Ambi"

# Boolean
is_student = True

print("Age:", age)
print("Monthly Budget:", monthly_budget)
print("Student Name:", student_name)
print("Is Student:", is_student)

print("\nData Types:")
print(type(age))
print(type(monthly_budget))
print(type(student_name))
print(type(is_student))
```

```python
# List of daily expenses
expenses = [100, 250, 150, 500, 300, 200]

# Print the third item
print("All Expenses:", expenses)
print("Third Expense:", expenses[2])
```

```python
 # Student financial information
student_finance = {
    "name": "Ambi",
    "monthly_income": 10000,
    "monthly_expenses": 6500,
    "savings": 3500,
    "currency": "INR"
}

print("Student Finance Details:")

for key, value in student_finance.items():
    print(key, ":", value)
```

```python
# Iterate through the expense list

print("Daily Expenses:")

for expense in expenses:
    print("Expense Amount: ₹", expense)
```

```python
# Check spending based on the expense amount

expense = 2500

if expense > 5000:
    print("High Spending")

elif expense > 2000:
    print("Moderate Spending")

else:
    print("Low Spending")
```

```python
# Function to calculate remaining savings

def calculate_savings(income, expenses):
    savings = income - expenses
    return savings

# Function call
result = calculate_savings(10000, 6500)

print("Remaining Savings: ₹", result)
```

```python
print("===== FINANCE EXPENSE ANALYSIS =====")

income = 10000
expenses = [100, 250, 150, 500, 300]

total_expenses = sum(expenses)
savings = income - total_expenses

print("Monthly Income: ₹", income)
print("Total Expenses: ₹", total_expenses)
print("Remaining Savings: ₹", savings)

if savings > 5000:
    print("Financial Status: Good Savings")
elif savings > 0:
    print("Financial Status: Positive Savings")
else:
    print("Financial Status: Overspending")
```
